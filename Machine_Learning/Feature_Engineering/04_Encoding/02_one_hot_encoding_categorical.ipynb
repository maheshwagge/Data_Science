# One Hot Encoding – Interview Notes

---

## 🔹 What is One Hot Encoding?

**One Hot Encoding (OHE)** converts categorical variables into binary columns (0/1), where each unique category becomes a separate feature.

### Example

If we have:

Color = [Red, Blue, Green]

After One Hot Encoding:

| Color | Red | Blue | Green |
|-------|-----|------|-------|
| Red   | 1   | 0    | 0     |
| Blue  | 0   | 1    | 0     |
| Green | 0   | 0    | 1     |

---

## 🔹 Why is it Needed?

- Machine learning algorithms work with **numerical data**, not text.
- Prevents models from assuming **ordinal relationships**.

⚠️ Example of a problem with Label Encoding:

If:
- Red = 1  
- Blue = 2  

The model may assume **Blue > Red**, which is incorrect for nominal categories.

---

## 🔹 When is One Hot Encoding Useful?

✅ **Nominal categorical data (no inherent order)**  
- Colors  
- Countries  
- Cities  

✅ **Linear models**
- Logistic Regression  
- SVM  

✅ **Tree-based models (low cardinality)**
- Random Forest  
- XGBoost  

---

## 🔹 When NOT to Use One Hot Encoding?

❌ **High cardinality features**  
- Creates too many columns  
- Increases memory usage  

❌ **Ordinal data (has natural order)**  
Use **Ordinal Encoding** instead  
Example:
High School < Bachelor < Master  

❌ **Severe memory constraints**  
- OHE produces sparse matrices  

---

## 🔹 Importance of One Hot Encoding

- Eliminates false ordinal relationships  
- Preserves full category information  
- Works well with distance-based algorithms  
- Makes features independent

## One Hot Encoding - Complete Code Examples


### 1. Basic One Hot Encoding with Pandas

````python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split

# Sample data
df = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'red', 'blue'],
    'size': ['S', 'M', 'L', 'M', 'S'],
    'price': [100, 150, 200, 120, 130]
})

print("Original DataFrame:")
print(df)
print("\n")

# Method 1: pandas get_dummies (simplest)
df_encoded = pd.get_dummies(df, columns=['color', 'size'])
print("With get_dummies:")
print(df_encoded)
````

### 2. Handling Test Data Consistency


````python
# Train-test split
X = df[['color', 'size', 'price']]
y = [1, 0, 1, 0, 1]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Problem: Test data might have categories not in train
X_test.loc[X_test.index[0], 'color'] = 'yellow'  # Introducing unknown category

# Solution 1: get_dummies with alignment
train_encoded = pd.get_dummies(X_train, columns=['color', 'size'])
test_encoded = pd.get_dummies(X_test, columns=['color', 'size'])

# Align test columns with train
test_encoded = test_encoded.reindex(columns=train_encoded.columns, fill_value=0)
print("Train columns:", train_encoded.columns.tolist())
print("Test columns (aligned):", test_encoded.columns.tolist())
````

### 3. Sklearn OneHotEncoder (Production-Ready)

````python
# Initialize encoder
encoder = OneHotEncoder(
    sparse_output=False,  # Return dense array
    handle_unknown='ignore',  # Handle unknown categories
    drop='first'  # Avoid dummy variable trap
)

# Fit on training data
encoder.fit(X_train[['color', 'size']])

# Transform both sets
train_encoded = encoder.transform(X_train[['color', 'size']])
test_encoded = encoder.transform(X_test[['color', 'size']])

# Get feature names
feature_names = encoder.get_feature_names_out(['color', 'size'])
print("Encoded feature names:", feature_names)

# Combine with numerical features
train_final = np.hstack([train_encoded, X_train[['price']].values])
test_final = np.hstack([test_encoded, X_test[['price']].values])
````

### 4. ColumnTransformer for Mixed Types

````python
from sklearn.preprocessing import StandardScaler

# Define column types
categorical_cols = ['color', 'size']
numerical_cols = ['price']

# Create preprocessor
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_cols),
        ('cat', OneHotEncoder(
            handle_unknown='ignore',
            drop='first'
        ), categorical_cols)
    ])

# Fit and transform
X_processed = preprocessor.fit_transform(X_train)
print("Processed shape:", X_processed.shape)
````

### 5. Handling High Cardinality


````python
# Create data with many categories
np.random.seed(42)
n_samples = 1000
n_categories = 500

df_high_card = pd.DataFrame({
    'category': np.random.choice([f'cat_{i}' for i in range(n_categories)], n_samples),
    'value': np.random.randn(n_samples)
})

# Method 1: Keep top K categories
top_k = 50
top_categories = df_high_card['category'].value_counts().nlargest(top_k).index
df_high_card['category_top'] = df_high_card['category'].where(
    df_high_card['category'].isin(top_categories), 'other'
)

encoded_topk = pd.get_dummies(df_high_card, columns=['category_top'])
print(f"Original categories: {n_categories}")
print(f"After top-{top_k}: {encoded_topk.shape[1]-1} features")

# Method 2: Frequency encoding (alternative to OneHot)
freq_encoding = df_high_card['category'].value_counts(normalize=True)
df_high_card['category_freq'] = df_high_card['category'].map(freq_encoding)
print("\nFrequency encoding example:")
print(df_high_card[['category', 'category_freq']].head())
````

### 6. Avoiding Dummy Variable Trap

````python
# Demonstrate multicollinearity
df_small = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'red']
})

# Without dropping (creates multicollinearity)
dummies_all = pd.get_dummies(df_small, columns=['color'])
print("All dummies (multicollinear):")
print(dummies_all)
print("Sum of color columns:", dummies_all[['color_red', 'color_blue', 'color_green']].sum(axis=1))

# With dropping (avoids trap)
dummies_dropped = pd.get_dummies(df_small, columns=['color'], drop_first=True)
print("\nWith drop_first:")
print(dummies_dropped)
````

### 7. Memory Optimization with Sparse Matrices

````python
from scipy.sparse import csr_matrix

# Create sparse one-hot encoding
encoder_sparse = OneHotEncoder(
    sparse_output=True,  # Return sparse matrix
    handle_unknown='ignore'
)

# Large dataset simulation
df_large = pd.DataFrame({
    'user_id': np.random.choice([f'user_{i}' for i in range(10000)], 100000),
    'item_id': np.random.choice([f'item_{i}' for i in range(5000)], 100000)
})

# Encode
encoded_sparse = encoder_sparse.fit_transform(df_large[['user_id', 'item_id']])
print(f"Sparse matrix shape: {encoded_sparse.shape}")
print(f"Memory usage: {encoded_sparse.data.nbytes / 1e6:.2f} MB")

# Convert sparse to dense
encoded_dense = encoded_sparse.toarray()

print(f"Sparse matrix shape: {encoded_sparse.shape}")

# Sparse memory usage
sparse_memory = (
    encoded_sparse.data.nbytes +
    encoded_sparse.indptr.nbytes +
    encoded_sparse.indices.nbytes
) / 1e6

print(f"Sparse memory usage: {sparse_memory:.2f} MB")

# Dense memory usage
dense_memory = encoded_dense.nbytes / 1e6
print(f"Dense memory usage: {dense_memory:.2f} MB")
````

### 8. Complete Pipeline Example


````python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Create pipeline
pipeline = Pipeline([
    ('preprocessor', ColumnTransformer([
        ('num', StandardScaler(), ['price']),
        ('cat', OneHotEncoder(
            handle_unknown='ignore',
            sparse_output=False
        ), ['color', 'size'])
    ])),
    ('classifier', RandomForestClassifier(random_state=42))
])

# Train and evaluate
pipeline.fit(X_train, y_train)
y_pred = pipeline.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, y_pred):.3f}")

# Save pipeline for production
import joblib
joblib.dump(pipeline, 'onehot_pipeline.pkl')
````

### 9. Custom OneHotEncoder with Frequency Threshold


````python
class CustomOneHotEncoder:
    def __init__(self, min_freq=0.05):
        self.min_freq = min_freq
        self.categories_ = {}
        
    def fit(self, X, col):
        # Calculate frequencies
        freq = X[col].value_counts(normalize=True)
        # Keep categories above threshold
        self.categories_[col] = freq[freq >= self.min_freq].index.tolist()
        return self
    
    def transform(self, X, col):
        # Replace rare categories with 'other'
        X_transformed = X[col].copy()
        mask = ~X_transformed.isin(self.categories_[col])
        X_transformed.loc[mask] = 'other'
        
        # One-hot encode
        return pd.get_dummies(X_transformed, prefix=col)

# Usage
encoder_custom = CustomOneHotEncoder(min_freq=0.2)
encoder_custom.fit(df_high_card, 'category')
df_encoded_custom = encoder_custom.transform(df_high_card, 'category')
print("Custom encoding result:")
print(df_encoded_custom.head())
````

### 10. Common Pitfalls and Solutions

````python
# Pitfall 1: Encoding target variable
df_wrong = pd.DataFrame({
    'feature': ['A', 'B', 'C'],
    'target': ['yes', 'no', 'yes']
})

# WRONG: One-hot encode target
wrong_encoded = pd.get_dummies(df_wrong, columns=['target'])  # Don't do this!

# RIGHT: Use LabelEncoder for target
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df_wrong['target_encoded'] = le.fit_transform(df_wrong['target'])

# Pitfall 2: Data leakage
# WRONG: Fit on full data before split
encoder_leaky = OneHotEncoder()
encoder_leaky.fit(df[['color']])  # Don't do this!

# RIGHT: Fit only on training data
X_train, X_test = train_test_split(df, test_size=0.2, random_state=42)
encoder_correct = OneHotEncoder()
encoder_correct.fit(X_train[['color']])

# Pitfall 3: Forgetting to handle NaN values
df_with_nan = df.copy()
df_with_nan.loc[0, 'color'] = np.nan

# Handle NaN before encoding
df_with_nan['color'] = df_with_nan['color'].fillna('missing')
encoded_clean = pd.get_dummies(df_with_nan, columns=['color'])
print("\nHandling NaN values:")
print(encoded_clean)
````
